# Chapter 2: Vector and Matrix Transposition in NumPy and PyTorch

This notebook demonstrates how to perform **vector and matrix transposition** using both **NumPy** and **PyTorch**.

### Key Concepts:
- **Transpose ($T$):** Converts rows into columns and columns into rows.
- **Vector Transpose:** Transforms a row vector $(1 \times N)$ into a column vector $(N \times 1)$ or vice-versa.
- **Matrix Transpose:** Transforms an $M \times N$ matrix into an $N \times M$ matrix where the $i$-th row becomes the $i$-th column.
- **Double Transpose:** Transposing an array/tensor twice restores its original shape and orientation: $(A^T)^T = A$.

```python
# =====================================================================
# Step 0: Import Required Libraries
# =====================================================================
import numpy as np
import torch

print(f"NumPy Version: {np.__version__}")
print(f"PyTorch Version: {torch.__version__}")
```

--- 
## 1. Vector and Matrix Transposition in NumPy

```python
# =====================================================================
# 1.1 NumPy Vector Transposition
# =====================================================================

# Create a 2D NumPy array representing a row vector (shape: 1 x 4)
nv = np.array([[5, -1, 9, 2]])
print("Original NumPy Row Vector (shape:", nv.shape, "):")
print(nv)

# Transpose the row vector into a column vector (shape: 4 x 1)
# Using the succinct `.T` property on the ndarray object
nv_T = nv.T
print("\nTransposed NumPy Column Vector (shape:", nv_T.shape, "):")
print(nv_T)

# Double transpose: transposing back yields the original row vector
nv_TT = nv_T.T
print("\nDouble Transposed NumPy Vector (shape:", nv_TT.shape, "):")
print(nv_TT)
```

```python
# =====================================================================
# 1.2 NumPy Matrix Transposition
# =====================================================================

# Create a 2 x 4 matrix
nM = np.array([
    [3,  8, 1, 6],
    [4, -2, 7, 0]
])
print("Original NumPy Matrix (shape:", nM.shape, "):")
print(nM)

# Transpose the matrix: row 0 becomes column 0, row 1 becomes column 1
# Shape changes from (2 x 4) to (4 x 2)
nM_T = nM.T
print("\nTransposed NumPy Matrix (shape:", nM_T.shape, "):")
print(nM_T)

# Double transpose restores original matrix dimensions
nM_TT = nM_T.T
print("\nDouble Transposed NumPy Matrix (shape:", nM_TT.shape, "):")
print(nM_TT)
```

--- 
## 2. Vector and Matrix Transposition in PyTorch

```python
# =====================================================================
# 2.1 PyTorch Vector Transposition
# =====================================================================

# Create a PyTorch tensor row vector (shape: 1 x 4)
tv = torch.tensor([[5, -1, 9, 2]])
print("Original PyTorch Row Tensor (shape:", tv.shape, "):")
print(tv)

# Transpose tensor using the `.T` attribute
tv_T = tv.T
print("\nTransposed PyTorch Column Tensor (shape:", tv_T.shape, "):")
print(tv_T)

# Double transpose restores original shape
tv_TT = tv_T.T
print("\nDouble Transposed PyTorch Tensor (shape:", tv_TT.shape, "):")
print(tv_TT)
```

```python
# =====================================================================
# 2.2 PyTorch Matrix Transposition
# =====================================================================

# Create a 2 x 4 PyTorch tensor matrix
tM = torch.tensor([
    [3,  8, 1, 6],
    [4, -2, 7, 0]
])
print("Original PyTorch Matrix Tensor (shape:", tM.shape, "):")
print(tM)

# Transpose the tensor matrix
tM_T = tM.T
print("\nTransposed PyTorch Matrix Tensor (shape:", tM_T.shape, "):")
print(tM_T)

# Double transpose
tM_TT = tM_T.T
print("\nDouble Transposed PyTorch Matrix Tensor (shape:", tM_TT.shape, "):")
print(tM_TT)
```

## 3. Comparing Data Types Across Frameworks

```python
# =====================================================================
# 3. Verifying Object Types
# =====================================================================

print("Type of NumPy matrix variable (nM):      ", type(nM))
print("Type of NumPy transposed matrix (nM_T):   ", type(nM_T))
print("Type of PyTorch matrix variable (tM):    ", type(tM))
print("Type of PyTorch transposed matrix (tM_T):", type(tM_T))
```
